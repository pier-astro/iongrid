# Inspect iongrid and OGIP tables

Run this notebook from its directory after `python -m pip install -e ".[inspect]"` in the iongrid repo. The first cell creates small synthetic examples. Sliders select **stored grid nodes**; they do not interpolate.


```python
%matplotlib widget
from pathlib import Path

from iongrid.inspect import TableInspector

%run make_demo.py
```

## Iongrid cube: optical depth

The cube holds its original optical-depth values on the energy-bin centres.


```python
cube = TableInspector(Path("products/demo-tau.fits"))
cube.show(emin=4, emax=10)
```

## OGIP table: transmission

The derived multiplicative table holds `exp(-tau)` at the same nodes. Replace the path with one of your own `.amod`, `.mmod`, `.emod`, or iongrid `.fits` files.


```python
table = TableInspector(Path("products/demo-transmission.mmod"))
table.show(emin=4, emax=10)
```
