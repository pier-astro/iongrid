# Standalone photoionization runs

Run either engine cell with its executable configured.

```python
import sys
sys.dont_write_bytecode = True

from pathlib import Path
from tempfile import mkdtemp

import matplotlib.pyplot as plt
import numpy as np

from iongrid.cloudy import CloudyPoint
from iongrid.xstar import XstarPoint

root = Path(mkdtemp(prefix="iongrid-points-"))
params = dict(lognh=21, logxi=4, vturb=100, logdensity=10, logluminosity=46, gamma=1.8)
```

```python
point = CloudyPoint(**params)
directory = root / "cloudy"
point.prepare(directory)  # Optional: template=Path("cloudy.in") or text=deck
result = point.run(directory, executable="cloudy")
assert result["status"] == "completed", directory
spectrum = point.collect(directory)

with np.load(spectrum) as data:
    plt.semilogx(data["energy"], data["tau_total"])
plt.xlabel("Energy (keV)")
plt.ylabel("Optical depth");
```

```python
point = XstarPoint(**params)
directory = root / "xstar"
point.prepare(directory)  # Optional: template=Path("xstar.par") or text=deck
result = point.run(directory, executable="xstar")
assert result["status"] == "completed", directory
spectrum = point.collect(directory)

with np.load(spectrum) as data:
    plt.semilogx(data["energy"], data["tau_total"])
plt.xlabel("Energy (keV)")
plt.ylabel("Optical depth");
```
